<b> <a>  INTRODUCTION

Les plateformes numériques telles que YouTube génèrent quotidiennement un volume important de commentaires exprimant les opinions, réactions et impressions des utilisateurs. L’analyse automatique de ces commentaires représente un enjeu important pour les entreprises, les créateurs de contenu et les organisations souhaitant mieux comprendre la perception de leur audience.

Dans ce contexte, ce projet porte sur l’analyse des sentiments exprimés dans des commentaires publiés sur YouTube à propos de films proposés par une chaîne. Chaque commentaire est associé à une catégorie de sentiment : positif, négatif ou neutre.

L’objectif est de mettre en place une approche de Natural Language Processing (NLP) combinée à des techniques de Machine Learning afin de permettre la classification automatique des commentaires selon leur sentiment.

Plusieurs modèles d’apprentissage automatique seront développés, optimisés et comparés afin d’identifier celui qui offre les meilleures performances pour cette tâche de classification.

<b> <a>  PROBLEME METIER  

Les entreprises, plateformes et créateurs de contenu disposent d’un grand nombre de commentaires provenant de leurs utilisateurs. L’analyse manuelle de ces commentaires devient rapidement difficile lorsque le volume de données augmente.

Il est donc nécessaire de disposer d’un système capable d’identifier automatiquement si un commentaire exprime une opinion positive, négative ou neutre.

Un tel système peut permettre de :

- Mesurer la satisfaction et l’insatisfaction des utilisateurs 
- Identifier rapidement les opinions négatives 
- Comprendre les réactions du public à un contenu 
- Suivre l’évolution de la perception des utilisateurs 
- Automatiser l’analyse de grandes quantités de commentaires 
- Faciliter la prise de décision à partir des opinions exprimées.

Le problème métier consiste donc à automatiser l’identification du sentiment exprimé dans les commentaires YouTube afin de transformer des données textuelles non structurées en informations exploitables.

<b> <a>  OBJECTIF DU PROJET

L’objectif principal de ce projet est de développer un modèle de Machine Learning capable de classifier automatiquement les commentaires YouTube selon leur sentiment.

Plus précisément, le projet vise à :

- Explorer et comprendre le jeu de données textuelles 
- Nettoyer et préparer les commentaires ;
- Transformer les textes en représentations numériques exploitables par les modèles 
- Développer plusieurs modèles de classification 
- Optimiser leurs hyperparamètres 
- Evaluer leurs performances à l’aide de différentes métriques 
- Comparer les modèles obtenus 
- Sélectionner le modèle le plus performant pour la classification des sentiments.

Les modèles étudiés sont notamment :

- Logistic Regression 
- Multinomial Naive Bayes 
- Support Vector Classifier (SVC) 
- Random Forest Classifier.

Les performances seront notamment évaluées à l’aide de l’accuracy, la précision, le rappel et le score F1, afin d'obtenir une comparaison complète des différents modèles.

<b> <a>  PROBLEMATIQUE DATA SCIENCE  

La problématique d’analyse de données peut être formulée ainsi :

Comment exploiter les commentaires YouTube afin de construire un modèle de Machine Learning capable de classer automatiquement les opinions exprimées en trois catégories : positive, négative et neutre ?

Cette problématique peut être déclinée en plusieurs questions :

- Comment préparer et nettoyer efficacement les commentaires textuels ?
- Quelles techniques permettent de transformer les commentaires en données numériques exploitables par les algorithmes de Machine Learning ?
- Quel modèle de classification permet d'obtenir les meilleures performances ?
- Comment optimiser les hyperparamètres des différents modèles ?
- Quels sont les modèles les plus performants pour distinguer les sentiments positifs, négatifs et neutres ?
- Le modèle sélectionné est-il suffisamment performant pour être utilisé sur de nouveaux commentaires ?
