<b> <a>  DESCRIPTION DES DONNEES

Le jeu de données est constitué de commentaires publiés sur YouTube concernant des films proposés par une chaîne.

Chaque observation contient principalement :

- Comment :	Commentaire textuel publié par l'utilisateur

- Sentiment : Catégorie de sentiment associée au commentaire

La variable Sentiment constitue la variable cible du projet et comporte trois modalités :

- Positive
- Negative
- Neutral


le dataset contient :

- 18408 lignes (18408 commentaires et sentimens)
